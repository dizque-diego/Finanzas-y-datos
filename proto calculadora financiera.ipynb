```python
print("Bienvenido a la calculadora de Costos!")
print("Un fondo cobra comision todos los años sobre el valor total que tienes invertido")

inv = int(input("por favor ingrese el valor a invertir"))
rendimiento = float(input("por favor ingrese el rendimiento en numero decimal"))
ter = float(input("Ingrese el valor del TER"))
anios = int(input("ingrese el numero de años que durará la inversion"))
def inversion(inv, rendimiento, ter, anios):
    inversion_final = inv * (1 + rendimiento - ter) ** anios
    return inversion_final
    
inversion(inv, rendimiento, ter, anios)
```
