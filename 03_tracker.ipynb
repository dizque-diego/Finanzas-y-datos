```python
import yfinance as yf


```

```python
# Tracker de inversiones en USD

portafolio_arq = {"NVDA":0.11 , "BTC-USD": 0.00012, "VOO": 0.081, "PEPT":2}
total_usd = 0
valores_usd = {}
for ticker, cantidad in portafolio_arq.items():
    precio = yf.Ticker(ticker).history(period="1d")["Close"].iloc[-1]
    inv_act = precio * cantidad
    print(f"Tengo de {ticker} la cantidad de {cantidad} participaciones, es decir a hoy tengo US$ {inv_act:,.2f} en este activo")
    total_usd += inv_act
    valores_usd[ticker] = inv_act


for ticker, valor in valores_usd.items():
    perce = (valor / total_usd)*100
    print (f"de mi portafolio actual el activo {ticker} ocupa el {perce:.1f} %")
    

print(f"el valor actual de mi portafolio en USD es de {total_usd:,.2f}")   

    
```

```python
# Tracker de inversiones en COP 
portafolio_trii = {"ECOPETROL.CL":1471 , "PFCIBEST.CL": 2 , "PFGRUPOARG.CL": 2, "NUCO.CL":2}
total_cop = 0
valores_cop = {}
for ticker, cantidad in portafolio_trii.items():
    precio = yf.Ticker(ticker).history(period="1d")["Close"].iloc[-1]
    inv_act_cop = precio * cantidad
    print(f"Tengo de {ticker} la cantidad de {cantidad} participaciones, es decir a hoy tengo COP$ {inv_act_cop:,.2f} en este activo")
    total_cop += inv_act_cop
    valores_cop[ticker] = inv_act_cop

for ticker, valor in valores_cop.items():
    perce = (valor / total_cop) * 100
    print (f"de mi portafolio actual el activo {ticker} ocupa el {perce:.1f} %")
print(f"el valor actual de mi portafolio en COP es de {total_cop:,.0f}")   
```

```python
#Consolidacion de inversión en COP y en USD

trm_cop = yf.Ticker("COP=X").history(period="1d")["Close"].iloc[-1]
total_usd_to_cop = total_usd * trm_cop
print (f"$1 USD equivale a {trm_cop:,.0f} COP")
print(f"El total del portafolio en USD convertido a COP son: {total_usd_to_cop:,.0f} pesos")
total_inv_cop = total_usd_to_cop + total_cop
print(f"El valor total de mi portafolio corresponde a COP ${total_inv_cop:,.0f} pesos")
total_inv_usd = total_inv_cop / trm_cop
print(f"El valor total de mi portafolio corresponde a US ${total_inv_usd:,.2f} dolares")
```
