```python
#Calculo montecarlo del rendimiento de un ETF
import numpy as np

rendimiento = 10
volatilidad = 16
vola_norm = volatilidad/100
rendi_norm = rendimiento/100
def montecarlo (rendimiento,volatilidad):
    rend_ajs = (np.log(rendimiento+1)-(volatilidad**2)/2)
    resultado = np.random.normal (rend_ajs,volatilidad,(10000, 10))
    fact = np.exp(resultado)
    valores_finales = np.prod(fact,axis=1) * 100
    peor = np.min(valores_finales)
    mejor = np.max(valores_finales)
    prom = np.mean(valores_finales)
    percent = np.mean(valores_finales < 100)
    porciento = percent * 100
        
    return peor, mejor, prom, porciento 
montecarlo(rendi_norm,vola_norm)
```
